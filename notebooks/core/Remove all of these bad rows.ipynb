```python
import pandas as pd

df = pd.read_csv('Heart_disease.csv')

print("Before cleaning:", df.shape)   # (70000, 11)

# The 2 bad columns are index 3 and 4
# Column 3: ap_hi  (SBP) — keep only 80–200 mmHg
# Column 4: ap_lo  (DBP) — keep only 40–130 mmHg

sbp_col = 'ap_hi(Systolic blood pressure)'
dbp_col = 'ap_lo(Diastolic blood pressure)'

# See every bad row before removing
bad_rows = df[~df[sbp_col].between(80, 200) | ~df[dbp_col].between(40, 130)]
print("Bad rows to remove:", len(bad_rows))        # 1347
print("Bad row indices:", list(bad_rows.index))    # full list

# Remove them
df_clean = df[df[sbp_col].between(80, 200) & df[dbp_col].between(40, 130)]

print("After cleaning:", df_clean.shape)   # (68653, 11)
df_clean.to_csv('Heart_disease_clean.csv', index=False)
```
